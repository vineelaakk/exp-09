```python
import pandas as pd

# Create DataFrames
df1 = pd.DataFrame({'Name':['A','B'], 'Marks':[80,90]})
df2 = pd.DataFrame({'Name':['C','D'], 'Marks':[70,85]})

# Combine
print("Combine:")
print(pd.concat([df1,df2]))

# Merge
a = pd.DataFrame({'ID':[1,2], 'Name':['A','B']})
b = pd.DataFrame({'ID':[1,2], 'Marks':[80,90]})

print("Merge:")
print(pd.merge(a,b,on='ID'))

# Hierarchical Indexing
df1 = df1.set_index(['Name'])
print("Index:")
print(df1)

# Reshape
print("Reshape:")
print(pd.melt(df2))

# Pivot
df = pd.DataFrame({
    'Name':['A','A','B','B'],
    'Subject':['Math','Science','Math','Science'],
    'Marks':[80,85,90,95]
})

print("Pivot:")
print(df.pivot(index='Name',columns='Subject',values='Marks'))
```
